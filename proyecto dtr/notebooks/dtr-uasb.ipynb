## Introducción

La Distribución de Tiempos de Residencia (DTR) es una herramienta fundamental para caracterizar el flujo en reactores químicos...

## Metodología

Se utilizó un trazador de rodamina para analizar el comportamiento del flujo dentro del reactor.  
Los datos fueron procesados de Ecxel en Python utilizando librerías como pandas, numpy y matplotlib.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# -------------------------
# 1. Verificar ruta actual (opcional pero útil)
# -------------------------
print("Carpeta actual:", os.getcwd())

# -------------------------
# 2. Ruta del archivo
# -------------------------
file = "../data/datos_3Tapas_sinLodo.xlsx"

# -------------------------
# 3. Cargar archivo
# -------------------------
df = pd.read_excel(file, sheet_name="TRH5hP4", header=10)

# -------------------------
# 4. Limpiar columnas
# -------------------------
df.columns = df.columns.str.strip()

print("\nColumnas disponibles:")
print(df.columns)

# -------------------------
# 5. Seleccionar variables
# -------------------------
df_clean = df[["(TRH,min)", "C/Co"]].copy()
df_clean.columns = ["tiempo", "C"]

# -------------------------
# 6. LIMPIEZA (clave)
# -------------------------
df_clean["tiempo"] = pd.to_numeric(df_clean["tiempo"], errors="coerce")
df_clean["C"] = pd.to_numeric(df_clean["C"], errors="coerce")

df_clean = df_clean.dropna()
df_clean = df_clean.sort_values("tiempo")

print("\nDatos limpios:")
print(df_clean.head())

# -------------------------
# 7. Graficar C(t)
# -------------------------
plt.figure()
plt.plot(df_clean["tiempo"], df_clean["C"])
plt.xlabel("Tiempo (min)")
plt.ylabel("C/Co")
plt.title("Curva del trazador C(t)")
plt.grid()
plt.show()

# -------------------------
# 8. Calcular E(t)
# -------------------------
area = np.trapz(df_clean["C"], df_clean["tiempo"])
df_clean["E_t"] = df_clean["C"] / area

# -------------------------
# 9. Graficar E(t)
# -------------------------
plt.figure()
plt.plot(df_clean["tiempo"], df_clean["E_t"])
plt.xlabel("Tiempo (min)")
plt.ylabel("E(t)")
plt.title("Distribución de tiempos de residencia")
plt.grid()
plt.show()

# -------------------------
# 10. Tiempo medio
# -------------------------
t_m = np.trapz(df_clean["tiempo"] * df_clean["E_t"], df_clean["tiempo"])
print("\nTiempo medio de residencia:", t_m)

# -------------------------
# 11. Varianza
# -------------------------
varianza = np.trapz((df_clean["tiempo"] - t_m)**2 * df_clean["E_t"], df_clean["tiempo"])
print("Varianza:", varianza)

# -------------------------
# 12. Curva F(t)
# -------------------------
df_clean["F_t"] = np.cumsum(df_clean["E_t"] * np.gradient(df_clean["tiempo"]))

plt.figure()
plt.plot(df_clean["tiempo"], df_clean["F_t"])
plt.xlabel("Tiempo (min)")
plt.ylabel("F(t)")
plt.title("Función acumulada de la DTR")
plt.grid()
plt.show()

```python
## Interpretación de la función E(t)

La función E(t) corresponde a la distribución de tiempos de residencia (DTR), la cual describe la probabilidad de que una partícula permanezca cierto tiempo dentro del reactor.

Observaciones:
- La forma de E(t) permite evaluar la desviación respecto a un reactor ideal.
- Una distribución ancha indica dispersión y mezcla.
- Una distribución estrecha indica comportamiento cercano a flujo pistón.
```

```python
## Tiempo medio de residencia

El tiempo medio de residencia calculado representa el tiempo promedio que las partículas permanecen dentro del reactor.

Este valor puede compararse con el tiempo de retención hidráulico teórico para evaluar la eficiencia del sistema.
```

```python
## Varianza

La varianza mide la dispersión de los tiempos de residencia respecto al valor medio.

- Una varianza baja indica flujo más uniforme.
- Una varianza alta indica mayor mezcla y dispersión.

Este parámetro es clave para diagnosticar desviaciones del comportamiento ideal del reactor.
```

```python
## Función acumulada F(t)

La función F(t) representa la fracción acumulada del fluido que ha salido del reactor en función del tiempo.

Esta curva permite visualizar:
- El tiempo en el que comienza la salida del trazador
- La rapidez con la que el sistema alcanza el vaciado completo
```

## Conclusiones

El análisis de la distribución de tiempos de residencia (DTR) permitió evaluar el comportamiento hidráulico del reactor bajo diferentes condiciones de operación.

Se observó que el incremento del caudal en un reactor operando en régimen laminar no genera un impacto significativo en el patrón de flujo, siempre que los parámetros hidráulicos se mantengan constantes. Esto indica que el caudal no es el factor dominante en la dinámica del fluido dentro del reactor en estas condiciones.

Por otro lado, la configuración de las entradas mostró un efecto considerable sobre el comportamiento del sistema. Cuando las entradas se encuentran muy próximas entre sí, se favorece la formación de zonas muertas o regiones de baja circulación, lo cual puede disminuir la eficiencia del proceso y afectar el mezclado dentro del reactor.

En contraste, el aumento en el número de entradas contribuye a una mejor distribución del flujo, promoviendo un mezclado más homogéneo en el sistema. Esto resalta la importancia del diseño hidráulico en el desempeño del reactor.

Finalmente, las configuraciones 3EX y 3ET2 presentaron los mejores resultados en términos de distribución del flujo y eficiencia hidráulica, por lo que se consideran las más adecuadas dentro de las condiciones analizadas.
